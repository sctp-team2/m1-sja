```python
from feature_engineering import build_features

df_feature = build_features(df)   # df_raw stays untouched
```

```python
df_feature.head()
```

```python
## Validation: check if the new features are added correctly

# 1. Column count
print(f"Columns: {df_feature.shape[1]} (expected 67)")

# 2. Any unexpected NaNs in derived columns?
derived_cols = [c for c in df.columns if c not in df_feature.columns]
nan_summary = df[derived_cols].isna().sum()
print("\nNaN counts in derived columns (>0 only):")
print(nan_summary[nan_summary > 0].sort_values(ascending=False))

# 3. Salary outlier check (flags the spread issue)
print("\nTop 10 average_salary values:")
print(df_feature.nlargest(10, "average_salary")[
    ["postedCompany_name", "title", "average_salary", "salary_minimum", "salary_maximum"]
])

# 4. Quality flag prevalence — these should be sensible percentages
flag_cols = ["salary_undisclosed_flag", "salary_suspicious_low", "mass_hiring_flag",
             "zero_engagement_flag", "seniority_mismatch_flag", "is_reposted",
             "is_agency", "hidden_gem_flag", "low_engagement_flag"]
print("\nFlag prevalence (% of rows):")
print((df_feature[flag_cols].mean() * 100).round(1).sort_values(ascending=False))

# 5. Composite scores should be roughly z-distributed (mean~0, std~1)
score_cols = ["hard_to_fill_score", "demand_intensity_score", "posting_quality_score"]
print("\nComposite score summary (should be ~mean 0, std ~1):")
print(df_feature[score_cols].describe().loc[["mean", "std", "min", "max"]].round(3))

# 6. Within-category z-scores should also be ~z-distributed per category
print("\nSalary z-score by category (sample of 5):")
print(df_feature.groupby("category_1", observed=True)["salary_zscore_within_category"]
        .agg(["mean", "std"]).round(3).head())

# 7. Top hard-to-fill — sanity check by eye
print("\nTop 5 hard-to-fill roles:")
print(df_feature.nlargest(5, "hard_to_fill_score")[
    ["title", "category_1", "posting_duration_days", "metadata_repostCount",
     "applications_per_vacancy", "hard_to_fill_score"]
])
```
